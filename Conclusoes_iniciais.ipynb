📋 Detalhamento de Cada Categoria

- Abaixo está a explicação do que cada perfil faz, seus nomes inteiros (corrigindo os cortes do gráfico) e o valor representado:
1. NAN (Não Informado / Ausente)

    Nome Inteiro: Not a Number (Valor Ausente / Sem Cadastro).

    O que faz: Representa linhas da planilha da CONAB onde o campo TIPO DE ORG ficou em branco ou não foi preenchido na fonte original.

    Valor: R$ 119.207.500,00 (é o maior volume da base).

2. AGRICULTOR FAMILIAR / AGRICULTORES FAMILIARES

    Nome Inteiro: Agricultor(es) Familiar(es).

    O que faz: Produtores rurais de pequena escala cuja gestão e força de trabalho são majoritariamente da própria família. Produzem a maior parte dos produtos hortifrúti, grãos, laticínios e tubérculos consumidos no país.

    Valor: R$ 75.018.500,00 (soma do singular R$ 37,5M + plural R$ 37,4M).

3. ASSENTADO DA REFORMA AGRÁRIA / ASSENTADOS DA REFORMA AGRARIA

    Nome Inteiro: Assentado(s) da Reforma Agrária.

    O que faz: Famílias trabalhadoras rurais que receberam lotes de terra por meio do programa de reforma agrária (organizadas em assentamentos do INCRA/Estados). Produzem alimentos diversos de forma cooperativa ou individual.

    Valor: R$ 32.245.200,00 (soma do singular R$ 17,4M + plural R$ 14,8M).

4. PESCADORES ARTESANAIS / PESCADOR ARTESANAL

    Nome Inteiro: Pescador(es) Artesanal(is).

    O que faz: Trabalhadores que praticam a pesca comercial em pequena escala (marítima, fluvial ou lacustre) de forma direta, sem uso de grandes embarcações industriais. Fornecem peixes e frutos do mar frescos para o programa.

    Valor: R$ 9.704.600,00 (soma de R$ 7,3M + R$ 2,3M).

5. AGROEXTRATIVISTA

    Nome Inteiro: Agroextrativista.

    O que faz: Comunidades e produtores que combinam a agricultura familiar com a coleta sustentável de produtos da floresta e da biodiversidade local (ex.: açaí, castanha-do-pará, babassu, umbu, borracha).

    Valor: R$ 788.200,00.

6. QUILOMBOLA

    Nome Inteiro: Quilombola (Comunidades Remanescentes de Quilombos).

    O que faz: Produtores pertencentes a territórios étnico-raciais quilombolas. Praticam agricultura tradicional, de subsistência e comunitária, preservando cultivares crioulas e técnicas de cultivo sustentáveis.

    Valor: R$ 725.500,00.

7. COMUNIDADE INDÍGENA / POVOS INDÍGENAS

    Nome Inteiro: Comunidade Indígena / Povos Indígenas.

    O que faz: Organizações e associações formadas por povos originários que produzem alimentos tradicionais (mandioca, milho, banana, castanhas) dentro de suas terras demarcadas.

    Valor: R$ 725.500,00 (soma de R$ 560k + R$ 165,5k).
